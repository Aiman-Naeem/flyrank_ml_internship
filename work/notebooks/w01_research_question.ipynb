## 2. The question: decision, action, cost of a wrong call

thousands a content reviewer should look at first for a titleand meta rewrite since reviewers don't have time to manually check every page's CTR against what's normal for its position. A content or SEO reviewer acts on the ranked list rewriting the title and meta description of flagged pages. The cost of a wrong recommendation depends on which kind of mistake it is: a false positive might flag a page that didn't actually need fixing. It carries the bigger risk since an unnecessary rewrite could accidentally hurt a page that was already performing fine. A false negative might miss a page that really did need fixing. It is a lesser but real cost since it means a genuine improvement opportunity goes unrealized.

```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")
vis = df[(df['impressions_90d'] >= 500) & df['position_tier'].notna() & df['ctr'].notna()]
bench = vis.groupby('position_tier')['ctr'].median().rename('ctr_bench')
vis = vis.join(bench, on='position_tier')
flagged = vis['ctr'] < (vis['ctr_bench'] - 1.0)
print(f"{flagged.sum()} flagged out of {len(vis)} visible pages")
```

## 4. Careful words: what I can and can't claim

#### What my work can say:
- Which pages have a CTR notably below what's typical for other pages at the same position tier. It is an observed, decision-support signal not a diagnosis.
- a flagged page is worth a reviewer's attention for its title and meta description.

#### What my work cannot say:
- It cannot prove that rewriting the title will actually improve clicks. There will be no experiment was run to test that.
- It cannot explain why a page's CTR is low. The dataset doesn't contain title or meta description text so a low CTR could come from a weak title, mismatched search intent, low sample volume or something else entirely. The ranking flags the pattern, not the cause.
- It does not predict Google's search rankings or how its algorithm works. It's based only on already-observed CTR and position data, not any understanding of ranking factors.
